## 0.1 The story: Contoso relaunches a tent

Contoso Outdoors, a made-up retailer of camping and hiking gear, is changing how it works with AI, one job at a time.

Marketing is relaunching the **TrailMaster X4** tent, and the website needs a refresh, including a home for **TrailMate**, the support chatbot customers will talk to.

Here's the tent, straight from its product manual in the shared [Contoso Retail dataset](../../data/contoso-retail/README.md) (`data/contoso-retail/manuals/trailmaster-x4-tent.md`):

| | TrailMaster X4 Tent |
| --- | --- |
| **Brand** | OutdoorLiving |
| **Price** | $250 |
| **Sleeps** | 4 people |
| **Season** | 3-season |
| **In the bag** | Tent body, aluminum poles, rainfly, stakes, reflective guy lines, carry bag |
| **Weight** | 12 lbs |
| **Warranty** | 2-year limited |

> 🔎 **Look closely:** the manual says *"water-resistant construction to withstand light rain"* and, a few lines later, *"Waterproof: Yes"*. Which one goes in an ad? Promising more than the product does is called **overclaiming**, and it costs customers' trust. In this workshop we say *water-resistant, with a rainfly*, and nothing stronger.

## 0.2 What does a launch need?

A launch is a lot of separate jobs. Each one needs a different kind of model. Microsoft AI (MAI) builds its own models for several of them:

| The launch needs… | The job | MAI model family | Notebook |

## 0.3 Best result, or fastest result?

Most MAI models come in two versions: a **standard** one and a **Flash** one. Flash does the same job, faster and cheaper.

So the question we ask again and again is: *which of our jobs actually need the big one?*

| Pick Flash when… | Pick the standard model when… |
| --- | --- |
| You need lots of results | It's the hero result people will study |
| Someone is waiting, like on a phone line | Quality is the whole point, like a radio ad |
| You're exploring and drafting | It's the final version |

We won't take that table on trust. Every notebook measures both versions on the same job, and you decide.

## 0.6 Is my setup ready?

This cell doesn't call Azure. It only checks which settings are in place, so you know which notebooks you can run live. It prints ✅ or ⚠️, never the values themselves.

> 🔮 **Guess first:** if nothing is deployed, which notebooks can you still read?

```python
import os
import shutil
from pathlib import Path

from dotenv import load_dotenv

ROOT = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / "data" / "contoso-retail").is_dir())
HERE = ROOT / "workshops" / "mai-multimodal-product-launch"
load_dotenv(HERE / ".env")


def has(name: str) -> bool:
    value = os.environ.get(name, "").strip()
    return bool(value) and "<" not in value


mode = os.environ.get("DEMO_MODE", "replay").strip().lower() or "replay"
print(f"Mode in this folder's .env: {'🔁 replay: reading saved results' if mode == 'replay' else '🎨 live: calling the models'}")
print("Each notebook can override it in its own mode cell.\n")

# Voice and transcription model names have defaults in 02 and 03, so only the endpoint is required there.
checks = {
    "01 · Build the launch kit": ["FOUNDRY_PROJECT_ENDPOINT", "MODEL_IMAGE_HERO", "MODEL_IMAGE_BULK"],
    "02 · Give it a voice": ["FOUNDRY_PROJECT_ENDPOINT"],
    "03 · Hear your customers": ["FOUNDRY_PROJECT_ENDPOINT"],
}
for notebook, names in checks.items():
    missing = [name for name in names if not has(name)]
    saved = len([p for p in (HERE / "examples" / notebook[:2]).glob("*") if p.suffix != ".json"])
    live = "✅ ready to run live" if not missing else f"⚠️ live needs {', '.join(missing)}"
    print(f"{notebook:28} {live:46} · 🔁 {saved} saved results to read")

node = shutil.which("node")
print(f"{'04 · Refresh the storefront':28} {'✅ Node.js found' if node else '⚠️ live needs Node.js':46} · plus GitHub Copilot in VS Code")
```

## 0.7 Ground rules for every notebook

These keep the workshop honest.

| Rule | What it means here |
| --- | --- |
| **Show, don't tell** | Every section starts with a customer's question, and answers it by running the model. No feature lists. |
| **Judge with a checklist** | Each result comes with *what to look for*. Score it yourself, rather than trusting one lucky run. |
| **Every run is different** | Your pictures and audio won't match ours. That's normal. |
| **Be honest about limits** | Each section names what slipped in our run. |
| **Check the release status** | MAI-Image-2.6 and Flash are generally available in Foundry. MAI-Voice-2.1 and Flash are in preview, and the MAI-Code Flash model in Copilot may be too. |
| **Say when it's AI** | Tell people when a voice or a picture was made by AI. See the [text-to-speech transparency note](https://learn.microsoft.com/azure/ai-foundry/responsible-ai/speech-service/text-to-speech/transparency-note) and [content credentials for images](https://aka.ms/FoundryProvenanceLearn). |
